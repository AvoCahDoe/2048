# 02 — DQN training curves

Double DQN on 2048: episode score, max tile, ε, and TD loss.

```bash
uv run twenty48-train --config experiments/configs/dqn_default.yaml
```

````python
from pathlib import Path
import json
import pandas as pd
from twenty48.viz import load_metrics_jsonl, plot_learning_curves, plot_ablation_grid

ROOT = Path("..").resolve()
run = ROOT / "experiments/runs/dqn_default"
metrics = load_metrics_jsonl(run / "metrics.jsonl")
metrics.tail()
````

````python
plot_learning_curves(metrics, window=40).show()
summary = json.loads((run / "summary.json").read_text())
summary
````

## Ablations
Compare `dqn_default`, `dqn_lr_low`, `dqn_tile_bonus`, `dqn_wide` under `experiments/runs/`.

````python
runs_dir = ROOT / "experiments/runs"
summaries = {}
for p in sorted(runs_dir.iterdir()):
    sp = p / "summary.json"
    if sp.exists() and p.name.startswith("dqn"):
        summaries[p.name] = json.loads(sp.read_text())
if summaries:
    plot_ablation_grid(summaries).show()
    display(pd.DataFrame([{"run": k, **v} for k, v in summaries.items()]))
else:
    print("No DQN summaries yet — train configs under experiments/configs/")
````
