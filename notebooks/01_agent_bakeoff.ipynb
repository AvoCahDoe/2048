# 01 — Agent bakeoff

Monte Carlo comparison of **random**, **heuristic**, and **expectimax** on classic 2048.

Run from repo root after `uv sync --extra dev`:
```bash
uv run twenty48-sim --bakeoff --episodes 30 --depth 1
uv run jupyter lab notebooks/01_agent_bakeoff.ipynb
```

````python
from pathlib import Path
import pandas as pd
from twenty48.viz import (
    plot_score_distributions,
    plot_max_tile_cdf,
    plot_move_direction_hist,
    plot_tile_emergence,
)

ROOT = Path("..").resolve()
csv_path = ROOT / "experiments/runs/bakeoff/bakeoff_all.csv"
df = pd.read_csv(csv_path)
df.head()
````

````python
summary = (
    df.groupby("agent")[["score", "max_tile", "steps", "won"]]
    .agg(["mean", "median", "max", "std"])
    .round(2)
)
summary
````

````python
plot_score_distributions(df).show()
plot_max_tile_cdf(df).show()
plot_tile_emergence(df).show()
for agent in sorted(df["agent"].unique()):
    plot_move_direction_hist(df, agent).show()
````

## Interpretation
- Random establishes a weak baseline (low max tiles).
- Heuristic uses empty/smoothness/snake weights — strong cheap policy.
- Expectimax (depth 1) looks one spawn ahead; deeper search improves play at higher compute cost.
